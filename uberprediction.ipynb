```python
import pandas as pd
```

```python
dataframe=pd.read_csv('uber.csv')
```

```python
df=dataframe
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.drop(['Unnamed: 0','key'],axis=1,inplace=True)
```

```python
df.pickup_datetime=pd.to_datetime(df['pickup_datetime'],errors='coerce')
```

```python
df.info()
```

```python

import matplotlib.pyplot as plt
plt.boxplot(df["fare_amount"])
```

```python
def find_outliers(df):
  q1=df.quantile(0.25)
  q3=df.quantile(0.75)
  IQR=q3-q1
  outliers=df[(df<q1-1.5*IQR)|(df>q3+1.5*IQR)]
  return outliers
```

```python
fare_amount_outliers=find_outliers(df['fare_amount'])
print('No. of Outliers: ',len(fare_amount_outliers))
print('Max Outlier: ',fare_amount_outliers.max())
print('Min Outlier: ',fare_amount_outliers.min())
```

```python
q1=df['fare_amount'].quantile(0.25)
q3=df['fare_amount'].quantile(0.75)
IQR=q3-q1
df=df[(df['fare_amount']>q1-1.5*IQR) & (df['fare_amount']<q3+1.5*IQR)]

```

```python
df.drop(df[df['fare_amount'] < 0].index, inplace = True)
```

```python
x=df.drop('fare_amount',axis=1)
y=df['fare_amount']
```

```python
x['pickup_datetime']=pd.to_numeric(pd.to_datetime(df['pickup_datetime']))
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=1)
#
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
lr_pred=lr.predict(x_test)
lr_pred
```

```python
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
import numpy as np
print("Mean Absolute Error: ",mean_absolute_error(y_test,lr_pred))
print("Mean Squared Error: ",mean_squared_error(y_test,lr_pred))
print("Root Mean Squared Error: ",np.sqrt(mean_squared_error(y_test,lr_pred)))
print("R2 Score: ",r2_score(y_test,lr_pred))
```

```python
from sklearn.ensemble import RandomForestRegressor
rf=RandomForestRegressor(n_estimators=100,random_state=101)
rf.fit(x_train,y_train)
```

```python
rf_pred = rf.predict(x_test)
rf_pred
```

```python
print("Mean Absolute Error: ",mean_absolute_error(y_test,rf_pred))
print("Mean Squared Error: ",mean_squared_error(y_test,rf_pred))
print("Root Mean Squared Error: ",np.sqrt(mean_squared_error(y_test,rf_pred)))
print("R2 Score: ",r2_score(y_test,rf_pred))
```
